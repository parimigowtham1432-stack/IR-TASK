```python
# ============================================================
# TF-IDF + VECTOR SPACE MODEL
# 300 NEWS DOCUMENTS
# GOOGLE COLAB COMPLETE SOURCE CODE
# ============================================================

# CELL 1: INSTALL LIBRARIES

!pip -q install pandas numpy scikit-learn matplotlib seaborn


# CELL 2: IMPORT LIBRARIES

import os
import re
import zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
from sklearn.feature_extraction.text import TfidfVectorizer, ENGLISH_STOP_WORDS
from sklearn.metrics.pairwise import cosine_similarity

os.makedirs("data", exist_ok=True)
os.makedirs("results", exist_ok=True)

RANDOM_STATE = 42


# CELL 3: GENERATE 300-DOCUMENT DATASET

categories = {
    "Technology": [
        "artificial intelligence", "machine learning",
        "cybersecurity", "cloud computing", "software",
        "robotics", "data analytics", "computer processor"
    ],
    "Sports": [
        "cricket tournament", "football match",
        "basketball team", "championship", "athlete training",
        "stadium", "sports league", "coach"
    ],
    "Business": [
        "stock market", "company revenue",
        "business investment", "banking sector",
        "economic growth", "financial report",
        "industry trade", "market demand"
    ],
    "Politics": [
        "government policy", "national election",
        "parliament session", "public services",
        "political campaign", "minister announcement",
        "voter participation", "law reform"
    ],
    "Entertainment": [
        "new film", "music concert",
        "television series", "movie director",
        "film actor", "streaming platform",
        "audience review", "entertainment industry"
    ]
}

templates = [
    "The latest {topic} development attracted attention after experts shared a detailed report.",
    "Officials and analysts discussed {topic} and its possible impact on the public.",
    "A new initiative involving {topic} was announced this week by leading organizations.",
    "Researchers and professionals examined {topic} during an event attended by many people.",
    "The report about {topic} highlights recent changes, future plans, and important challenges.",
    "Several groups responded to the {topic} update and described its expected benefits.",
    "The development in {topic} may influence upcoming decisions across the sector.",
    "A detailed review of {topic} was published following recent events and discussions."
]

rng = np.random.default_rng(RANDOM_STATE)

rows = []
doc_id = 1

for category, topics in categories.items():
    for i in range(60):
        topic = topics[i % len(topics)]
        template = templates[(i + list(categories).index(category)) % len(templates)]

        title = f"{category} update: {topic.title()} report {i+1}"

        text = template.format(topic=topic)

        text += (
            f" The {category.lower()} sector continues to monitor "
            "new developments and public response."
        )

        rows.append({
            "ID": doc_id,
            "Title": title,
            "Text": text,
            "Category": category
        })

        doc_id += 1

df = pd.DataFrame(rows)

df.to_csv("data/news_dataset.csv", index=False)

print("Dataset generated successfully!")
print("Total documents:", len(df))
print(df["Category"].value_counts())


# CELL 4: OPTIONAL UPLOAD YOUR OWN DATASET

# If you already have a 300-record dataset, uncomment below.

# uploaded = files.upload()
# filename = list(uploaded.keys())[0]
# df = pd.read_csv(filename)

# Required columns: ID, Title, Text, Category


# CELL 5: DATASET ANALYSIS

print("\nDATASET INFORMATION")
print("-------------------")

print("Total documents:", len(df))
print("Total categories:", df["Category"].nunique())
print("\nCategory distribution:")

print(df["Category"].value_counts())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate records:", df.duplicated(subset=["Title", "Text"]).sum())

df = df.drop_duplicates(subset=["Title", "Text"]).reset_index(drop=True)

df["Combined_Text"] = (
    df["Title"].fillna("").astype(str)
    + " "
    + df["Text"].fillna("").astype(str)
)


# CELL 6: TEXT PREPROCESSING

STOP_WORDS = set(ENGLISH_STOP_WORDS)

def clean_text(text):

    text = str(text).lower()

    text = re.sub(r"https?://\S+|www\.\S+", " ", text)

    text = re.sub(r"<[^>]+>", " ", text)

    text = re.sub(r"[^a-z\s]", " ", text)

    tokens = text.split()

    tokens = [
        word for word in tokens
        if word not in STOP_WORDS and len(word) > 1
    ]

    return " ".join(tokens)


df["Cleaned_Text"] = df["Combined_Text"].apply(clean_text)

df.to_csv("results/cleaned_dataset.csv", index=False)

print("\nTEXT PREPROCESSING COMPLETED")

print("Original:")
print(df["Combined_Text"].iloc[0])

print("\nCleaned:")
print(df["Cleaned_Text"].iloc[0])


# CELL 7: BASIC TF-IDF

basic_vectorizer = TfidfVectorizer(
    lowercase=False,
    tokenizer=str.split,
    preprocessor=None,
    token_pattern=None,
    use_idf=True,
    smooth_idf=False,
    sublinear_tf=False,
    ngram_range=(1, 1),
    norm="l2"
)

basic_matrix = basic_vectorizer.fit_transform(
    df["Cleaned_Text"]
)

basic_terms = basic_vectorizer.get_feature_names_out()

print("\nBASIC TF-IDF")
print("-------------")
print("Documents:", basic_matrix.shape[0])
print("Vocabulary:", basic_matrix.shape[1])

basic_avg = np.asarray(basic_matrix.mean(axis=0)).ravel()

basic_top = pd.DataFrame({
    "Term": basic_terms,
    "Average_TFIDF": basic_avg
})

basic_top = basic_top.sort_values(
    "Average_TFIDF",
    ascending=False
)

display(basic_top.head(20))

basic_top.to_csv(
    "results/top_terms_basic.csv",
    index=False
)


# CELL 8: IMPROVED TF-IDF

improved_vectorizer = TfidfVectorizer(
    lowercase=False,
    tokenizer=str.split,
    preprocessor=None,
    token_pattern=None,
    use_idf=True,
    smooth_idf=True,
    sublinear_tf=True,
    ngram_range=(1, 2),
    norm="l2"
)

improved_matrix = improved_vectorizer.fit_transform(
    df["Cleaned_Text"]
)

improved_terms = improved_vectorizer.get_feature_names_out()

print("\nIMPROVED TF-IDF")
print("----------------")
print("Documents:", improved_matrix.shape[0])
print("Vocabulary:", improved_matrix.shape[1])

improved_avg = np.asarray(
    improved_matrix.mean(axis=0)
).ravel()

improved_top = pd.DataFrame({
    "Term": improved_terms,
    "Average_TFIDF": improved_avg
})

improved_top = improved_top.sort_values(
    "Average_TFIDF",
    ascending=False
)

display(improved_top.head(20))

improved_top.to_csv(
    "results/top_terms_improved.csv",
    index=False
)


# CELL 9: SAVE TF-IDF MATRICES

basic_dense = pd.DataFrame(
    basic_matrix.toarray(),
    columns=basic_terms
)

basic_dense.insert(0, "Document_ID", df["ID"])

basic_dense.to_csv(
    "results/tfidf_basic.csv",
    index=False
)

improved_dense = pd.DataFrame(
    improved_matrix.toarray(),
    columns=improved_terms
)

improved_dense.insert(0, "Document_ID", df["ID"])

improved_dense.to_csv(
    "results/tfidf_improved.csv",
    index=False
)

print("TF-IDF matrices saved successfully.")


# CELL 10: TOP 20 BASIC TF-IDF TERMS CHART

plt.figure(figsize=(10, 7))

sns.barplot(
    data=basic_top.head(20),
    x="Average_TFIDF",
    y="Term"
)

plt.title("Top 20 Basic TF-IDF Terms")
plt.tight_layout()

plt.savefig("results/top_terms_basic.png")

plt.show()


# CELL 11: TOP 20 IMPROVED TF-IDF TERMS CHART

plt.figure(figsize=(10, 7))

sns.barplot(
    data=improved_top.head(20),
    x="Average_TFIDF",
    y="Term"
)

plt.title("Top 20 Improved TF-IDF Terms")

plt.tight_layout()

plt.savefig("results/top_terms_improved.png")

plt.show()


# CELL 12: VECTOR SPACE MODEL

basic_similarity = cosine_similarity(basic_matrix)

improved_similarity = cosine_similarity(improved_matrix)

print("\nVECTOR SPACE MODEL")
print("------------------")

print("Basic similarity matrix:", basic_similarity.shape)

print("Improved similarity matrix:", improved_similarity.shape)


# CELL 13: SAVE SIMILARITY MATRIX

pd.DataFrame(
    basic_similarity,
    index=df["ID"],
    columns=df["ID"]
).to_csv("results/similarity_basic.csv")

pd.DataFrame(
    improved_similarity,
    index=df["ID"],
    columns=df["ID"]
).to_csv("results/similarity_improved.csv")


# CELL 14: SIMILARITY HEATMAP

plt.figure(figsize=(12, 9))

sns.heatmap(
    improved_similarity[:30, :30],
    cmap="YlGnBu"
)

plt.title("TF-IDF Cosine Similarity Heatmap")

plt.xlabel("Documents")

plt.ylabel("Documents")

plt.tight_layout()

plt.savefig("results/similarity_heatmap.png")

plt.show()


# CELL 15: FIND TOP 5 SIMILAR DOCUMENTS

def top_similar_documents(document_index, similarity_matrix, top_k=5):

    if not 0 <= document_index < len(df):
        raise IndexError("Invalid document index")

    scores = similarity_matrix[document_index].copy()

    scores[document_index] = -1

    indices = np.argsort(scores)[::-1][:top_k]

    result = df.iloc[indices][
        ["ID", "Title", "Category", "Text"]
    ].copy()

    result["Similarity_Score"] = scores[indices]

    return result.reset_index(drop=True)


print("\nSELECTED DOCUMENT")

display(
    df.iloc[[0]][
        ["ID", "Title", "Category", "Text"]
    ]
)

print("\nTOP 5 SIMILAR DOCUMENTS")

display(
    top_similar_documents(
        0,
        improved_similarity,
        top_k=5
    )
)


# CELL 16: CUSTOM TEXT SEARCH

def search_custom_text(
    query,
    vectorizer,
    document_matrix,
    top_k=5
):

    cleaned_query = clean_text(query)

    if not cleaned_query:
        raise ValueError("Enter valid text.")

    query_vector = vectorizer.transform(
        [cleaned_query]
    )

    scores = cosine_similarity(
        query_vector,
        document_matrix
    ).ravel()

    indices = np.argsort(scores)[::-1][:top_k]

    result = df.iloc[indices][
        ["ID", "Title", "Category", "Text"]
    ].copy()

    result["Similarity_Score"] = scores[indices]

    return result.reset_index(drop=True)


query = """
Artificial intelligence and machine learning
are transforming modern software technology.
"""

print("\nCUSTOM TEXT SEARCH")

display(
    search_custom_text(
        query,
        improved_vectorizer,
        improved_matrix,
        top_k=5
    )
)


# CELL 17: BASIC VS IMPROVED COMPARISON

comparison = []

for i in range(min(50, len(df) - 1)):

    j = i + 1

    comparison.append({

        "Document_A": df.loc[i, "ID"],

        "Document_B": df.loc[j, "ID"],

        "Basic_Similarity": basic_similarity[i, j],

        "Improved_Similarity": improved_similarity[i, j]

    })

comparison_df = pd.DataFrame(comparison)

display(comparison_df.head(20))

comparison_df.to_csv(
    "results/similarity_comparison.csv",
    index=False
)


# CELL 18: MODEL CONFIGURATION COMPARISON

version_summary = pd.DataFrame([

    {
        "Version": "Basic TF-IDF",
        "TF": "Linear",
        "IDF": "log(N/df)",
        "N-Gram": "Unigram",
        "Vocabulary": len(basic_terms)
    },

    {
        "Version": "Improved TF-IDF",
        "TF": "Sublinear",
        "IDF": "Smoothed IDF",
        "N-Gram": "Unigram + Bigram",
        "Vocabulary": len(improved_terms)
    }

])

display(version_summary)

version_summary.to_csv(
    "results/version_comparison.csv",
    index=False
)


# CELL 19: EXPORT RESULTS

zip_path = "TFIDF_VSM_Final_Results.zip"

with zipfile.ZipFile(
    zip_path,
    "w",
    zipfile.ZIP_DEFLATED
) as zip_file:

    for root, dirs, filenames in os.walk("results"):

        for filename in filenames:

            filepath = os.path.join(root, filename)

            zip_file.write(filepath)


print("\n================================")
print("TF-IDF PROJECT COMPLETED")
print("================================")

print("Total Documents:", len(df))

print("Basic Vocabulary:", len(basic_terms))

print("Improved Vocabulary:", len(improved_terms))

print("Results saved in:", zip_path)

files.download(zip_path)
```
